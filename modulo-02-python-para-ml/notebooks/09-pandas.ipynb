## DataFrames

Un DataFrame es una tabla 2D con filas y columnas etiquetadas. Se puede crear desde listas, diccionarios, o archivos CSV. Es la estructura central para trabajar con datos en ML.

````python
# Forma 1: desde listas + nombres de columnas
data = [
    ['Nissan', 'Stanza', 1991, 138, 4, 'MANUAL', 'sedan', 2000],
    ['Hyundai', 'Sonata', 2017, None, 4, 'AUTOMATIC', 'Sedan', 27150],
    ['Lotus', 'Elise', 2010, 218, 4, 'MANUAL', 'convertible', 54990],
    ['GMC', 'Acadia',  2017, 194, 4, 'AUTOMATIC', '4dr SUV', 34450],
    ['Nissan', 'Frontier', 2017, 261, 6, 'MANUAL', 'Pickup', 32340],
]

columns = [
    'Make', 'Model', 'Year', 'Engine HP', 'Engine Cylinders',
    'Transmission Type', 'Vehicle_Style', 'MSRP'
]
````

````python
df = pd.DataFrame(data, columns=columns)
````

## Series

Una Serie es una columna individual del DataFrame. Se accede con corchetes `df['columna']`. Si el nombre tiene espacios, NO se puede usar `df.columna`.

````python
# Esto da error porque el nombre tiene espacio
# La notación con punto solo funciona si el nombre NO tiene espacios
df["Model"].values
````

````python
df["Model"]
````

````python
# Forma correcta: usar corchetes con string
df['Engine HP']
````

````python
# Seleccionar varias columnas: pasar una lista de nombres
df[['Make', 'Model', 'MSRP']]
````

````python
# Crear una nueva columna: simplemente asignar con df['nombre_nuevo']
df['id'] = [1, 2, 3, 4, 5]
````

````python
# Sobrescribir la columna existente
df['id'] = [10, 20, 30, 40, 50]
````

````python
df
````

````python
# Eliminar una columna con del
del df['id']
````

## Índice

Cada fila tiene un índice (por defecto numérico, empezando en 0). Se puede cambiar a strings o reset.

````python
df.index
````

````python
df.Make.index
````

````python
# Cambiar el índice a letras
df.index = ['a', 'b', 'c', 'd', 'e']
````

````python
df
````

````python
# iloc: acceso por posición numérica (integer location)
# Seleccionar filas 1, 2 y 4
df.iloc[[1, 2, 4]]
````

````python
# Restaurar índice numérico: drop=True descarta el índice anterior
df.reset_index(drop=True, inplace=True)
````

## Acceso a elementos\n\nUsa `iloc[fila, columna]` para acceso por posición o `loc[etiqueta]` por etiqueta del índice.

## Operaciones elemento a elemento\n\nAl igual que NumPy, las operaciones se aplican a cada valor de la columna.

````python
# Multiplicar toda una columna por un escalar (los NaN se propagan)
df['Engine HP'] * 2
````

````python
# Comparación: devuelve Serie booleana (útil para filtrar)
df['Year'] >= 2015
````

## Filtrado

Pasamos una condición booleana dentro de `df[...]` para quedarnos solo con las filas que cumplen. Para combinar condiciones usamos `&` (AND) y `|` (OR), con paréntesis obligatorios.

````python
# Filtrar: solo autos Nissan
df[df['Make'] == 'Nissan']
````

````python
# Combinar condiciones: Nissan Y año >= 2015
# IMPORTANTE: usar & (no 'and') y paréntesis en cada condición
df[
    (df['Make'] == 'Nissan') & (df['Year'] >= 2015)
]
````

## Operaciones con strings

Pandas permite aplicar métodos de strings a columnas de texto usando `.str`. Muy útil para limpiar nombres de columnas o valores categóricos.

````python
# replace en un string de Python normal

'machine learning zoomcamp'.replace(' ', '_')
````

````python
# .str.lower() convierte toda la columna a minúsculas
df['Vehicle_Style'].str.lower()
````

````python
# Encadenar operaciones: reemplazar espacios por guiones bajos Y convertir a minúsculas
df['Vehicle_Style'] = df['Vehicle_Style'].str.replace(' ', '_').str.lower()
````

````python
df
````

## Estadísticas descriptivas

`describe()` genera un resumen estadístico de las columnas numéricas. `nunique()` muestra cuántos valores distintos tiene cada columna.

````python
# describe(): media, std, min, max, cuartiles de columnas numéricas
df.describe().round(2)
````

````python
# nunique(): número de valores únicos por columna
df.nunique()
````

## Valores nulos (NaN)

`isnull().sum()` muestra cuántos valores faltan por columna. Detectar y manejar nulos es un paso crítico en la preparación de datos.

````python
# Contar valores nulos por columna: Engine HP tiene 1 nulo
df.isnull().sum()
````

## Agrupamiento (groupby)

`groupby()` agrupa filas por una columna y aplica una función de agregación. Es el equivalente a `GROUP BY` en SQL.

# Equivalente SQL:

```sql
SELECT transmission_type, MAX(MSRP)
FROM cars
GROUP BY transmission_type
```

````python
# Precio máximo por tipo de transmisión
df.groupby('Transmission Type').MSRP.max()
````

## Conversión a arrays de NumPy

Para pasar datos a modelos de ML, necesitamos convertir columnas de Pandas a arrays de NumPy usando `.values`.

````python
# .values devuelve un array de NumPy
df.MSRP.values
````

````python
# to_dict: convierte el DataFrame a una lista de diccionarios (útil para APIs)
df.to_dict(orient='records')
````
