## Multiplicación

Hay tres tipos de multiplicación en álgebra lineal, cada una con reglas distintas sobre las dimensiones.

```python
# Dimensión del vector: ambos deben tener el mismo tamaño para el producto punto
v.shape[0]
```

```python
v.shape
```

```python
# Producto punto (dot product) implementado manualmente
# 
# Suma de u[i] * v[i] para cada posición i En ML, esto es como calcular: predicción = features · pesos = [w1, w2] w1 num de baños, w2 num de habitaciones = (w1, x1) + (w2, x2) + w0
def vector_vector_multiplication(u, v):
    assert u.shape[0] == v.shape[0]
    
    n = u.shape[0]
    
    result = 0.0
    
    for i in range(n):
        result = result + u[i] * v[i]
    
    return result
```

```python
# Nuestra implementación manual: u·v = 2*1 + 4*0 + 5*0 + 6*2 = 14
vector_vector_multiplication(u, v)
```

```python
# En la práctica usamos .dot() de NumPy (mucho más rápido)
u.dot(v)
```

```python
# Matriz U de 3x4: 3 muestras, 4 features cada una
U = np.array([
    [2, 4, 5, 6],
    [1, 2, 1, 2],
    [3, 1, 2, 1],
])
```

```python
# Shape: (filas, columnas) → U es 3x4
U.shape
```

```python
# Multiplicación matriz-vector: cada fila de U hace producto punto con v
# # Requisito: columnas de U == elementos de v
# Resultado: un vector con tantos elementos como filas tiene U
def matrix_vector_multiplication(U, v):
    assert U.shape[1] == v.shape[0]
    
    num_rows = U.shape[0]
    
    result = np.zeros(num_rows)
    
    for i in range(num_rows):
        result[i] = vector_vector_multiplication(U[i], v)
    
    return result
```

```python
# Resultado: vector de 3 elementos (una predicción por muestra)
matrix_vector_multiplication(U, v)
```

```python
# Equivalente con NumPy: U.dot(v) o U @ v
U.dot(v)
```

```python
# Matriz V de 4x3: necesaria para multiplicar con U (3x4)
# Regla: columnas de U (4) == filas de V (4)
V = np.array([
    [1, 1, 2],
    [0, 0.5, 1], 
    [0, 2, 1],
    [2, 1, 0],
])
```

```python
# Multiplicación matriz-matriz: aplica matriz-vector para cada columna de V
# U(3x4) · V(4x3) = Resultado(3x3)


# U es 3×4 → 3 muestras, 4 features
# V es 4×3 → 4 features, 3 "pesos/predicciones"
# Columnas de U (4) == Filas de V (4) ✅
# Resultado: 3×3 → filas de U × columnas de V


# U                          V                       R
# [2 4 5 6]   ·   [1  1   2 ]   =   [14  20  13 ]
# [1 2 1 2]       [0  0.5 1 ]       [ 5   6   5 ]
# [3 1 2 1]       [0  2   1 ]       [ 5  8.5  9 ]
#                 [2  1   0 ]


def matrix_matrix_multiplication(U, V):
    assert U.shape[1] == V.shape[0]
    
    num_rows = U.shape[0]
    num_cols = V.shape[1]
    
    result = np.zeros((num_rows, num_cols))
    
    for i in range(num_cols):
        vi = V[:, i]
        Uvi = matrix_vector_multiplication(U, vi)
        result[:, i] = Uvi
    
    return result
```

```python
# Nuestra implementación manual
matrix_matrix_multiplication(U, V)
```

```python
# Equivalente con NumPy (mismo resultado, mucho más rápido)
U.dot(V)
```

## Matriz identidad

La matriz identidad (I) es el equivalente al número 1 en multiplicación: cualquier matriz multiplicada por I da la misma matriz. Tiene unos en la diagonal y ceros en el resto.

```python
# Crear matriz identidad 3x3
I = np.eye(3)
I
```

```python
# V · I = V (la identidad no cambia nada)
V.dot(I)
```
