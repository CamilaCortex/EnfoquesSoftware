```python
# Crear nueva columna 'Price Range' para facilitar la visualización
def getrange(Price):
    if (Price >= 0 and Price < 25000):
        return '0 - 25000'
    if (Price >= 25000 and Price < 50000):
        return '25000 - 50000'
    if (Price >= 50000 and Price < 75000):
        return '50000 - 75000'
    if (Price >= 75000 and Price < 100000):
        return '75000 - 100000'

df['Price Range'] = df.apply(lambda x: getrange(x['Price']), axis=1)

df['Price Range'].value_counts()
```

```python
# Separar variables dependientes e independientes

x = df[['Popularity', 'Year', 'HP', 'Cylinders', 'MPG-H', 'MPG-C']].values
y = df['Price'].values

print(f"Dimensiones de X (características): {x.shape}")
print(f"Dimensiones de y (objetivo - precio): {y.shape}")
```

#### División del Conjunto de Datos
Típicamente, separamos un conjunto de datos en un conjunto de entrenamiento y un conjunto de prueba. La mayoría de los datos se usa para entrenamiento, y una porción más pequeña se usa para pruebas. Analysis Services muestrea aleatoriamente los datos para asegurar que los conjuntos de prueba y entrenamiento sean similares.

```python
# División del dataset en conjuntos de entrenamiento y prueba
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=0)

print(f"Entrenamiento - X: {x_train.shape}, y: {y_train.shape}")
print(f"Prueba        - X: {x_test.shape}, y: {y_test.shape}")
```

#### Regresión Lineal
La regresión lineal es un tipo básico y comúnmente usado de análisis predictivo. La idea general de la regresión es examinar dos cosas:

(1) ¿Un conjunto de variables predictoras hace un buen trabajo al predecir una variable de resultado (dependiente)?

(2) ¿Qué variables en particular son predictores significativos de la variable objetivo, y de qué manera —indicada por la magnitud y el signo de los estimadores beta— impactan la variable objetivo?

Estas estimaciones de regresión se usan para explicar la relación entre una variable dependiente y una o más variables independientes.

```python
from sklearn.linear_model import LinearRegression

lr_model = LinearRegression() # de acá sale su b0, resto de los b1, b2,... , b5 b, theta, w
lr_model.fit(x_train, y_train)

# Evaluación del modelo de Regresión Lineal
print("=== Regresión Lineal ===")
print(f"Precisión en entrenamiento : {lr_model.score(x_train, y_train):.4f}")
print(f"Precisión en prueba        : {lr_model.score(x_test, y_test):.4f}")
```

```python
print("Intercepto (β₀):", lr_model.intercept_)
print("Coeficientes (β₁...βₙ):", lr_model.coef_)
```

```python
features = ['Popularity', 'Year', 'HP', 'Cylinders', 'MPG-H', 'MPG-C']

print(f"Intercepto (β₀): {lr_model.intercept_[0]:.6f}")
for name, coef in zip(features, lr_model.coef_[0]):
    print(f"  {name:12s} → {coef:.6f}")
```

```python
y_pred = lr_model.predict(x_test)
y_pred[0:5]
```

```python
plt.scatter(y_test, y_pred)
plt.xlabel("Valores Reales")
plt.ylabel("Valores Predichos")
plt.title("Regresión Lineal: Valores Reales vs Predichos")
plt.show()
```

```python
sns.distplot((y_test-y_pred),bins=50)
```

```python
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
import math

print("=== Métricas de Evaluación - Regresión Lineal ===")
print(f"R² Score              : {r2_score(y_test, y_pred):.4f}")
print(f"Error Cuadrático Medio: {mean_squared_error(y_test, y_pred):.4f}")
print(f"Error Absoluto Medio  : {mean_absolute_error(y_test, y_pred):.4f}")
print(f"RMSE                  : {math.sqrt(mean_squared_error(y_test, y_pred)):.4f}")
```

### Validación de Supuestos — QQ-Plot de Residuos

El **QQ-Plot (Quantile-Quantile Plot)** compara los cuantiles de los residuos del modelo contra los cuantiles de una distribución normal teórica.

**¿Cómo interpretarlo?**
- **Puntos sobre la línea diagonal roja** → los errores siguen distribución normal ✅
- **Colas pesadas (puntos se alejan en los extremos)** → distribución con colas más largas que la normal (leptocúrtica)
- **Forma de "S"** → asimetría en los residuos

Un modelo de regresión lineal asume que los residuos son aproximadamente normales para que los intervalos de confianza y los p-valores sean válidos.

```python
import scipy.stats as stats

residuos = y_test - y_pred

fig, axes = plt.subplots(1, 2, figsize=(15, 6))

# QQ-Plot
stats.probplot(residuos.flatten(), dist="norm", plot=axes[0])
axes[0].set_title("QQ-Plot de Residuos — Regresión Lineal")
axes[0].get_lines()[0].set(color='steelblue', markersize=3, alpha=0.5)
axes[0].get_lines()[1].set(color='red', linewidth=2)

# Histograma de residuos para comparar
axes[1].hist(residuos.flatten(), bins=50, edgecolor='black', color='steelblue', alpha=0.7)
axes[1].set_title("Distribución de Residuos")
axes[1].set_xlabel("Residuo (y_real - y_pred)")
axes[1].set_ylabel("Frecuencia")

plt.tight_layout()
plt.show()

# Test estadístico de normalidad (Shapiro-Wilk sobre muestra de 2000 obs.)
muestra = residuos.flatten()[:2000]
stat, p_valor = stats.shapiro(muestra)
print(f"\nTest Shapiro-Wilk (muestra n=2000):")
print(f"  Estadístico W : {stat:.4f}")
print(f"  p-valor       : {p_valor:.6f}")
print(f"  Conclusión    : {'Residuos NO normales (p < 0.05)' if p_valor < 0.05 else 'No se rechaza normalidad (p >= 0.05)'}")
```

### Lo que ves en tu gráfico:

- Centro (cuantiles -1 a +1): los puntos siguen bien la línea roja → los errores pequeños/medianos sí se comportan normalmente ✅

- Cola derecha (cuantiles > 2): los puntos se disparan hacia arriba fuera de la línea → el modelo subestima 
sistemáticamente algunos precios altos (hay autos caros que el modelo predice muy por debajo del valor real), es decir recuerda partir que error = y_real - y_pred, si el y_pred es menor que y_real, entonces el error es positivo y se desvía hacia arriba (cola derecha), es decir modelo subestima precio real. 


- Cola izquierda (cuantiles < -2): también hay desviación, pero menor, en estos casos el modelo sobreestima precio (y_pred > y_real), es decir error negativo se desvía hacia abajo (cola izquierda), es decir modelo sobreestima precio real.
